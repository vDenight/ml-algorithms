```python
@beartype
def dot_product(
    features: Float[np.ndarray, "M K"],
    weights: Float[np.ndarray, "K N"]
) -> Float[np.ndarray, "M N"]:
    """
    Multiplies two matrices.
    The inner dimension (K) must match between both inputs.
    """
    return features @ weights
```
