```python
from trees.base_tree import BaseTree
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd

# 1. Load the data
print("Loading Iris dataset...")
iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

# Optional: Add the target labels (0, 1, 2) as a new column at the end
df['target'] = iris.target

# 2. Format to match your specific type hints
X = np.array(iris.data, dtype=np.float64)
labels = np.array(iris.target, dtype=np.uint64)

df.head(10)
```

```python
# 3. Initialize your custom tree
print("Training BaseTree (max_depth=5)...")
tree = BaseTree(max_depth=5)

# 4. Fit the data
tree.fit(X, labels)

# 5. Print the tree structure
print("\n--- Trained Decision Tree ---")
tree.print_tree()
```

```python
from sklearn.model_selection import train_test_split

X = np.array(iris.data, dtype=np.float64)
labels = np.array(iris.target, dtype=np.uint64)

X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)

# 4. Initialize your custom tree
print("Training BaseTree (max_depth=5)...")
tree = BaseTree(max_depth=5, min_samples_split_ratio=0.03, min_samples_leaf_ratio=0.01)

# 5. Fit the data on the TRAINING set only
tree.fit(X_train, y_train)

# 6. Print the tree structure
print("\n--- Trained Decision Tree ---")
tree.print_tree()

# 7. Evaluate accuracy on the TEST set
print("\n--- Evaluation ---")
predictions = tree.predict(X_test)
correct_predictions = np.sum(predictions == y_test)
total_predictions = len(y_test)
accuracy = correct_predictions / total_predictions
print(f"Test Accuracy: {accuracy:.2%} ({correct_predictions}/{total_predictions} correct)")
```

```python
from sklearn.model_selection import KFold


# Format to match your specific type hints
X = np.array(iris.data, dtype=np.float64)
labels = np.array(iris.target, dtype=np.uint64)

# 2. Set up K-Fold Cross Validation
k = 5
kf = KFold(n_splits=k, shuffle=True, random_state=42)

fold_accuracies = []

print(f"\nStarting {k}-Fold Cross Validation (max_depth=5)...")

# 3. Iterate through each fold
for fold, (train_idx, test_idx) in enumerate(kf.split(X), start=1):
    # Split the data using the indices provided by KFold
    X_train, y_train = X[train_idx], labels[train_idx]
    X_test, y_test = X[test_idx], labels[test_idx]

    # Initialize a fresh tree for this fold
    tree = BaseTree(max_depth=5, min_samples_split_ratio=0.03, min_samples_leaf_ratio=0.01)

    # Fit on the training fold
    tree.fit(X_train, y_train)

    # Evaluate on the test fold
    predictions = tree.predict(X_test)
    correct_predictions = np.sum(predictions == y_test)
    total_predictions = len(y_test)
    accuracy = correct_predictions / total_predictions
    fold_accuracies.append(accuracy)

    print(f"Fold {fold} Accuracy: {accuracy:.2%}")

# 4. Calculate the true average performance
average_accuracy = np.mean(fold_accuracies)
print(f"\nAverage Accuracy across {k} folds: {average_accuracy:.2%}")
```
